```python
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from pyswarms.single.global_best import GlobalBestPSO
from scipy.stats import pearsonr
```

```python
# -------------------------------
# 2. PDFS-PSO fitness function
# -------------------------------
def fitness_function(particle, X_data, y_data, alpha=0.85):
    n_particles = particle.shape[0]
    fitness = np.zeros(n_particles)

    for i in range(n_particles):
        mask = particle[i] > 0.5
        if np.sum(mask) < 2:  # minimum 2 features
            fitness[i] = 1e6
            continue

        X_selected = X_data.values[:, mask]
        clf = RandomForestClassifier(n_estimators=50, random_state=42)
        score = cross_val_score(clf, X_selected, y_data, cv=5, scoring='accuracy').mean()

        # Correlation penalty
        if X_selected.shape[1] > 1:
            corr_matrix = np.corrcoef(X_selected, rowvar=False)
            corr_matrix = np.abs(corr_matrix)
            np.fill_diagonal(corr_matrix, 0)
            high_corr = np.sum(corr_matrix > alpha)
        else:
            high_corr = 0

        fitness[i] = -score + 0.001 * high_corr
    return fitness
```

```python
# -------------------------------
# 3. Run PDFS-PSO for fixed alpha thresholds
# -------------------------------
thresholds = [0.85, 0.75, 0.65, 0.50]
results = {}

for alpha in thresholds:
    print(f"\nRunning PDFS-PSO for alpha = {alpha} ...")

    optimizer = GlobalBestPSO(n_particles=30, dimensions=X_train.shape[1], options={'c1':0.8,'c2':0.8,'w':0.7})
    best_cost, best_pos = optimizer.optimize(fitness_function, iters=25, X_data=X_train, y_data=y_train, alpha=alpha)

    selected_features = np.where(best_pos > 0.5)[0]
    results[alpha] = {
        'features': selected_features,
        'num_features': len(selected_features)
    }

    print(f"Alpha {alpha}: selected {len(selected_features)} features")
```

```python
from sklearn.model_selection import GridSearchCV, cross_val_score
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, confusion_matrix
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier, StackingClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
```

```python
# -------------------------------
# 4. Define classifiers and hyperparameter grids
# -------------------------------
models = {
    "LR": LogisticRegression(C=1.0, max_iter=1000),

    "SVM": SVC(
        C=1.0,
        kernel='rbf',
        gamma='scale',
        probability=True
    ),

    "RF": RandomForestClassifier(
        n_estimators=100,
        max_depth=None,
        n_jobs=-1,
        random_state=42
    ),

    "LGBM": LGBMClassifier(
        num_leaves=31,
        learning_rate=0.1,
        n_estimators=100,
        n_jobs=-1,
        random_state=42
    )
}
```

```python
from sklearn.metrics import (
    accuracy_score, recall_score, precision_score, f1_score,
    matthews_corrcoef, roc_auc_score, log_loss, confusion_matrix
)
```

```python
results = {}

for name, model in models.items():
    print(f"\n================ {name} =================")

    # Train model
    model.fit(X_train, y_train)

    # ---------------- TRAIN ----------------
    train_pred = model.predict(X_train)
    train_proba = model.predict_proba(X_train)[:, 1]

    tn, fp, fn, tp = confusion_matrix(y_train, train_pred).ravel()

    train_metrics = {
        "Accuracy": accuracy_score(y_train, train_pred),
        "Recall": recall_score(y_train, train_pred),
        "Precision": precision_score(y_train, train_pred),
        "F1-Score": f1_score(y_train, train_pred),
        "MCC": matthews_corrcoef(y_train, train_pred),
        "AUC": roc_auc_score(y_train, train_proba),
        "LogLoss": log_loss(y_train, train_proba),
        "TP": tp, "TN": tn, "FP": fp, "FN": fn
    }

    # ---------------- TEST ----------------
    test_pred = model.predict(X_test)
    test_proba = model.predict_proba(X_test)[:, 1]

    tn, fp, fn, tp = confusion_matrix(y_test, test_pred).ravel()

    test_metrics = {
        "Accuracy": accuracy_score(y_test, test_pred),
        "Recall": recall_score(y_test, test_pred),
        "Precision": precision_score(y_test, test_pred),
        "F1-Score": f1_score(y_test, test_pred),
        "MCC": matthews_corrcoef(y_test, test_pred),
        "AUC": roc_auc_score(y_test, test_proba),
        "LogLoss": log_loss(y_test, test_proba),
        "TP": tp, "TN": tn, "FP": fp, "FN": fn
    }

    # Store results
    results[name] = {
        "Train": train_metrics,
        "Test": test_metrics
    }

    # ---------------- DISPLAY ----------------
    print("---- Train Metrics ----")
    for k, v in train_metrics.items():
        print(f"{k}: {v:.4f}" if isinstance(v, float) else f"{k}: {v}")

    print("\n---- Test Metrics ----")
    for k, v in test_metrics.items():
        print(f"{k}: {v:.4f}" if isinstance(v, float) else f"{k}: {v}")
```

```python
rows = []
for model, data in results.items():
    row = {"Model": model}
    for metric, value in data["Test"].items():
        row[metric] = value
    rows.append(row)

results_df = pd.DataFrame(rows)
```
